```python
import pandas as pd
import pickle
from pathlib import Path

from config import DATA_DIR, PHEN_MANIFEST_PATH

phen_ids = pd.read_csv(DATA_DIR / 'phenotype_identifier.csv', usecols=[
    'phen_id',
    'filename'
    ])

phen_filenames = {}
for idx, row in phen_ids.iterrows():
    phen_filenames[row['phen_id']] = row['filename']

with open(Path(DATA_DIR) / "phen_filenames.pkl", "wb") as f:
    pickle.dump(phen_filenames, f)
```

```python
phen_ids = pd.read_csv(DATA_DIR / 'phenotype_identifier.csv', usecols=[
    'phen_id',
    'filename',
    'n_cases_EUR'
    ])

phen_cases = {}
for idx, row in phen_ids.iterrows():
    phen_cases[row['phen_id']] = row['n_cases_EUR']

with open(Path(DATA_DIR) / "phen_cases.pkl", "wb") as f:
    pickle.dump(phen_cases, f)
```
