### Generate a dictionary of reference file names by chromosome

```python
import re
import os
import json
from pathlib import Path

from config import REF_DIR_1KG, DATA_DIR
```

```python
chr_ref_dict = {}
```

```python


pattern = re.compile(r'\.MAF_threshold=0\.005\.(\d+)_combined\.csv')
csv_files = [f for f in os.listdir(REF_DIR_1KG) if f.endswith('.csv') and f != 'targets.csv']

for csv_file in csv_files:
    match = pattern.search(csv_file)
    chromosome = int(match.group(1))
    chr_ref_dict[chromosome] = csv_file
```

```python
chr_ref_dict = dict(sorted(chr_ref_dict.items()))
```

```python
chr_ref_dict
```

```python
dict_path = Path(DATA_DIR) / 'reference_files_by_chr.json'

with open(dict_path, 'w') as json_file:
    json.dump(chr_ref_dict, json_file)
```
